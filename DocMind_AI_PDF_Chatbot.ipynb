```python
from sentence_transformers import CrossEncoder

reranker = CrossEncoder("cross-encoder/ms-marco-MiniLM-L-6-v2")

print("Reranker loaded successfully!")
```

```python
def rerank_results(question, results):
    
    pairs = [
        [question, result["text"]]
        for result in results
    ]

    scores = reranker.predict(pairs)

    ranked_results = []

    for result, score in zip(results, scores):
        ranked_results.append({
            "page": result["page"],
            "text": result["text"],
            "score": float(score)
        })

    ranked_results.sort(
        key=lambda x: x["score"],
        reverse=True
    )

    return ranked_results
```

```python
results = search_pdf(
    "What are the main objectives?",
    k=5
)

reranked = rerank_results(
    "What are the main objectives?",
    results
)

for rank, result in enumerate(reranked, start=1):
    print(f"\n===== Reranked Result {rank} =====")
    print("Score:", result["score"])
    print("Page:", result["page"])
    print(result["text"][:500])
```

```python
def search_pdf_reranked(question, retrieve_k=5, final_k=2):

    # 1. Retrieve candidates using FAISS
    results = search_pdf(
        question,
        k=retrieve_k
    )

    # 2. Rerank candidates
    ranked_results = rerank_results(
        question,
        results
    )

    # 3. Keep only the best results
    final_results = ranked_results[:final_k]

    return final_results
```

```python
final_results = search_pdf_reranked(
    "What are the main objectives?",
    retrieve_k=5,
    final_k=2
)

for rank, result in enumerate(final_results, start=1):

    print(f"\n===== Final Result {rank} =====")
    print("Page:", result["page"])
    print("Score:", result["score"])
    print(result["text"])
```

```python
def ask_pdf(question, retrieve_k=5, final_k=2):

    # 1. Retrieve and rerank
    results = search_pdf_reranked(
        question,
        retrieve_k=retrieve_k,
        final_k=final_k
    )

    # 2. Build context
    context = ""

    for result in results:
        context += f"\n[Page {result['page']}]\n"
        context += result["text"]

    # 3. Create grounded prompt
    prompt = f"""
You are DocMind, an AI assistant that answers questions about a PDF.

RULES:
1. Use ONLY the information in the PDF context.
2. Do not use outside knowledge.
3. Answer the user's question directly.
4. Ignore chunks that only contain a question, heading, or test question
   if they do not provide the actual answer.
5. Do not invent information.
6. If the answer is not supported by the context, say:
   "I could not find the answer in the PDF."

PDF CONTEXT:
{context}

USER QUESTION:
{question}
"""

    # 4. Ask Ollama
    response = ollama.chat(
        model="llama3.2:3b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    answer = response["message"]["content"]

    # 5. Get source pages
    source_pages = sorted(
        set(result["page"] for result in results)
    )

    return answer, source_pages
```

```python
def get_supported_results(question, retrieve_k=5, threshold=1.0):

    # Retrieve and rerank
    ranked_results = search_pdf_reranked(
        question,
        retrieve_k=retrieve_k,
        final_k=retrieve_k
    )

    # Keep only sufficiently relevant chunks
    supported_results = [
        result
        for result in ranked_results
        if result["score"] >= threshold
    ]

    return supported_results
```

```python
def ask_pdf(question, retrieve_k=5, threshold=1.0):

    # 1. Retrieve and rerank
    ranked_results = search_pdf_reranked(
        question,
        retrieve_k=retrieve_k,
        final_k=retrieve_k
    )

    # 2. Keep only relevant evidence
    results = [
        result
        for result in ranked_results
        if result["score"] >= threshold
    ]

    # 3. Handle no relevant evidence
    if not results:
        return (
            "I could not find the answer in the PDF.",
            []
        )

    # 4. Build context
    context = ""

    for result in results:
        context += f"\n[Page {result['page']}]\n"
        context += result["text"]

    # 5. Grounded prompt
    prompt = f"""
You are DocMind, an AI assistant that answers questions about a PDF.

Rules:
1. Use ONLY the information in the PDF context.
2. Do not use outside knowledge.
3. Answer the user's question directly.
4. Do not invent information.
5. If the answer is not supported by the context, say:
   "I could not find the answer in the PDF."

PDF CONTEXT:
{context}

USER QUESTION:
{question}
"""

    # 6. Generate answer with Ollama
    response = ollama.chat(
        model="llama3.2:3b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    answer = response["message"]["content"]

    # 7. Source pages
    source_pages = sorted(
        set(result["page"] for result in results)
    )

    return answer, source_pages
```

```python
def rerank_structured_results(question, results):

    pairs = [
        [question, result["text"]]
        for result in results
    ]

    scores = reranker.predict(pairs)

    ranked_results = []

    for result, score in zip(results, scores):

        ranked_results.append({
            "page": result["page"],
            "section": result["section"],
            "text": result["text"],
            "distance": result["distance"],
            "score": float(score)
        })

    ranked_results.sort(
        key=lambda x: x["score"],
        reverse=True
    )

    return ranked_results
```

```python
def retrieve_relevant_chunks(
    question,
    retrieve_k=5,
    final_k=2,
    threshold=1.0
):

    # Step 1: FAISS retrieval
    results = search_structured_pdf(
        question,
        k=retrieve_k
    )

    # Step 2: Cross-Encoder reranking
    ranked_results = rerank_structured_results(
        question,
        results
    )

    # Step 3: Keep sufficiently relevant results
    filtered_results = [
        result
        for result in ranked_results
        if result["score"] >= threshold
    ]

    # Step 4: Keep only the best evidence
    final_results = filtered_results[:final_k]

    return final_results
```

```python
def identify_evidence(question, results):

    evidence_results = []

    for result in results:

        evidence_question = f"""
Question:
{question}

Document passage:
{result['text']}

Does this document passage directly contain the answer
to the question?

Answer only:
YES
or
NO
"""

        response = ollama.chat(
            model="llama3.2:3b",
            messages=[
                {
                    "role": "user",
                    "content": evidence_question
                }
            ]
        )

        decision = response["message"]["content"].strip().upper()

        if "YES" in decision:
            evidence_type = "Direct evidence"
        else:
            evidence_type = "Supporting evidence"

        evidence_results.append({
            "page": result["page"],
            "section": result["section"],
            "score": result["score"],
            "evidence_type": evidence_type,
            "text": result["text"]
        })

    return evidence_results
```

```python
def adaptive_retrieve(
    question,
    retrieve_k=5,
    max_k=3,
    score_gap=2.0,
    min_score=1.0
):

    results = search_structured_pdf(
        question,
        k=retrieve_k
    )

    ranked_results = rerank_structured_results(
        question,
        results
    )

    if not ranked_results:
        return []

    best_score = ranked_results[0]["score"]

    # If even the best result is below the minimum
    # relevance threshold, treat the question as unsupported.
    if best_score < min_score:
        return []

    selected = []

    for result in ranked_results:

        if len(selected) >= max_k:
            break

        # Two conditions:
        # 1. Result must be absolutely relevant.
        # 2. Result must be reasonably close to the best result.
        if (
            result["score"] >= min_score
            and
            best_score - result["score"] <= score_gap
        ):
            selected.append(result)

    return selected
```

```python
from pdf_processor import process_pdf
from rag import create_vector_index, adaptive_retrieve

chunks = process_pdf(
    "data/uploaded.pdf"
)

index, embeddings, filtered_chunks = create_vector_index(
    chunks
)

print("Original chunks:", len(chunks))
print("Indexed chunks:", len(filtered_chunks))
print("FAISS vectors:", index.ntotal)

results = adaptive_retrieve(
    "What database does the platform use?",
    index,
    filtered_chunks
)

print("\nRETRIEVED RESULTS:\n")

for result in results:
    print(
        f"Page {result['page']} | "
        f"Section {result['section']} | "
        f"Score {result['score']:.2f}"
    )
    print(result["text"])
    print("-" * 80)
```
